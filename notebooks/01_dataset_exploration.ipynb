```python
from datasets import load_dataset

dataset = load_dataset("SetFit/ag_news")

print(dataset)
```

```python
print(dataset["train"][0])
```

```python
print(dataset["train"].unique("label_text"))
```

```python
from collections import Counter

label_counts = Counter(dataset["train"]["label_text"])

for category, count in label_counts.items():
    print(f"{category}: {count}")
```

```python
text_lengths = [len(text.split()) for text in dataset["train"]["text"]]

print("Minimum words:", min(text_lengths))
print("Maximum words:", max(text_lengths))
print("Average words:", sum(text_lengths) / len(text_lengths))
```
